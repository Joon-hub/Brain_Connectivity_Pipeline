```python
import pandas as pd 
import numpy as np
import os
from pathlib import Path
```

```python
# import train data 
df = pd.read_csv("/home/sjoon/projects/brain_connectivity_classifier/data/raw/PIOP2_restingstate.csv")
df.shape
```

```python
# Get region names (assuming they're column names or you have a list)
regions = df.columns.tolist()  # or your region list

# Separate by hemisphere
left_regions = []
right_regions = []

for region in regions:
    # Cortical regions: LH_/RH_ prefix
    if region.startswith('LH_'):
        left_regions.append(region)
    elif region.startswith('RH_'):
        right_regions.append(region)
    # Subcortical regions: -lh/-rh suffix
    elif region.endswith('-lh'):
        left_regions.append(region)
    elif region.endswith('-rh'):
        right_regions.append(region)

print(f"Left hemisphere regions: {len(left_regions)}")
print(f"Right hemisphere regions: {len(right_regions)}")
```
