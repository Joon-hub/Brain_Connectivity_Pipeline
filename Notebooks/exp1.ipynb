```python
def parse_cortex_label(label):
    parts = label.split('_')
    # structure: ['7Networks', 'LH', 'Vis', '1'] or ['7Networks', 'RH', 'Default', 'PFC', '2']
    hemisphere = parts[1]
    network = parts[2]
    # handle sub-networks like Default_PFC or DorsAttn_Post
    if len(parts) > 4:
        network += '_' + parts[3]
    return pd.Series([hemisphere, network])

cortex_df = cortex_df.copy()
cortex_df[['hemisphere', 'network']] = cortex_df['label'].apply(parse_cortex_label)
```

```python
network_counts = cortex_df['network'].value_counts()
print(network_counts)
print("Total networks:", len(network_counts))
```
