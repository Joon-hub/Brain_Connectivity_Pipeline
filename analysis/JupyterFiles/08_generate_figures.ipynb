# 08 - Generate Publication Figures
Create final high-quality figures for thesis.

```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

# Set publication-quality style
plt.style.use('seaborn-v0_8-paper')
sns.set_context('paper', font_scale=1.5)
plt.rcParams['figure.dpi'] = 300
plt.rcParams['savefig.dpi'] = 300
plt.rcParams['font.family'] = 'sans-serif'

PROJECT_ROOT = Path.cwd().parent if 'analysis' in str(Path.cwd()) else Path.cwd()
ANALYSIS_DIR = PROJECT_ROOT / 'data' / 'analysis'
FIGURES_DIR = PROJECT_ROOT / 'figures' / 'publication'
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

## 1. Main Performance Figure

```python
# Load data
cv_perf = pd.read_csv(ANALYSIS_DIR / 'compiled_cv_performance.csv')
task_perf = pd.read_csv(ANALYSIS_DIR / 'compiled_task_performance.csv')

# Create figure
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# CV performance
ax = axes[0]
full_cv = cv_perf[cv_perf['Analysis'] == 'Full Connectivity']
x = np.arange(len(full_cv))
ax.bar(x, full_cv['Accuracy'], yerr=full_cv['Accuracy_Std'], 
       capsize=5, color=['#1f77b4', '#ff7f0e', '#2ca02c'], 
       edgecolor='black', linewidth=1.5)
ax.set_xticks(x)
ax.set_xticklabels(full_cv['Strategy'])
ax.set_ylabel('Accuracy')
ax.set_title('A) Cross-Validation Performance')
ax.set_ylim([0, 1])
ax.grid(axis='y', alpha=0.3)

# Task performance
ax = axes[1]
full_task = task_perf[task_perf['Analysis'] == 'Full Connectivity']
ax.bar(x, full_task['Accuracy'], 
       color=['#1f77b4', '#ff7f0e', '#2ca02c'], 
       edgecolor='black', linewidth=1.5)
ax.set_xticks(x)
ax.set_xticklabels(full_task['Strategy'])
ax.set_ylabel('Accuracy')
ax.set_title('B) Task Testing Performance')
ax.set_ylim([0, 1])
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig(FIGURES_DIR / 'figure1_main_performance.png', dpi=300, bbox_inches='tight')
plt.savefig(FIGURES_DIR / 'figure1_main_performance.pdf', bbox_inches='tight')
print("Saved Figure 1")
plt.show()
```

## 2. Reorganization Figure

```python
# Load reorganization data
reorg = pd.read_csv(ANALYSIS_DIR / 'reorganization_analysis.csv')
network_reorg = pd.read_csv(ANALYSIS_DIR / 'network_reorganization.csv', index_col=0)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Network reorganization
ax = axes[0]
means = network_reorg['Reorganization_Score_mean']
stds = network_reorg['Reorganization_Score_std']
colors = ['#d62728' if x > 0 else '#2ca02c' for x in means]
ax.bar(range(len(means)), means, yerr=stds, capsize=5, 
       color=colors, edgecolor='black', linewidth=1.5, alpha=0.8)
ax.axhline(0, color='black', linewidth=2)
ax.set_xticks(range(len(means)))
ax.set_xticklabels(network_reorg.index, rotation=45, ha='right')
ax.set_ylabel('Reorganization Score')
ax.set_title('A) Network-Level Reorganization')
ax.grid(axis='y', alpha=0.3)

# Distribution
ax = axes[1]
ax.hist(reorg['Reorganization_Score'], bins=40, 
        edgecolor='black', alpha=0.7, color='steelblue')
ax.axvline(0, color='red', linestyle='--', linewidth=2)
ax.set_xlabel('Reorganization Score')
ax.set_ylabel('Number of Regions')
ax.set_title('B) Distribution of Reorganization')
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig(FIGURES_DIR / 'figure2_reorganization.png', dpi=300, bbox_inches='tight')
plt.savefig(FIGURES_DIR / 'figure2_reorganization.pdf', bbox_inches='tight')
print("Saved Figure 2")
plt.show()
```

```python
print("\n" + "="*80)
print("ALL FIGURES GENERATED")
print("="*80)
print(f"\nPublication-quality figures saved to: {FIGURES_DIR}")
print("\nGenerated:")
print("  - figure1_main_performance.png/.pdf")
print("  - figure2_reorganization.png/.pdf")
```
