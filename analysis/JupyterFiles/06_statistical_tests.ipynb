# 06 - Statistical Tests
Comprehensive statistical validation of results.

```python
import numpy as np
import pandas as pd
from pathlib import Path
from scipy import stats
from scipy.stats import wilcoxon, friedmanchisquare, mannwhitneyu
import json

PROJECT_ROOT = Path('/home/sjoon/projects/brain_connectivity_classifier')
ANALYSIS_DIR = PROJECT_ROOT / 'data' / 'analysis'
```

## 1. Strategy Comparison Tests

```python
# Load data
network_perf = pd.read_csv(ANALYSIS_DIR / 'compiled_network_performance.csv')
cv_networks = network_perf[network_perf['Phase'] == 'CV'].copy()

# Friedman test (non-parametric repeated measures)
multi = cv_networks[cv_networks['Strategy'] == 'MULTINOMIAL']['Accuracy'].values
ovo = cv_networks[cv_networks['Strategy'] == 'OVO']['Accuracy'].values
ovr = cv_networks[cv_networks['Strategy'] == 'OVR']['Accuracy'].values

stat, p = friedmanchisquare(multi, ovo, ovr)
print(f"Friedman test: χ²={stat:.4f}, p={p:.4e}")
```

## 2. Save Statistical Results

```python
results = {'friedman_test': {'statistic': float(stat), 'p_value': float(p)}}
with open(ANALYSIS_DIR / 'statistical_tests.json', 'w') as f:
    json.dump(results, f, indent=2)
print("Statistical tests complete!")
```
